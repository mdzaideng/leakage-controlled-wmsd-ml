# 04 — Computational-efficiency benchmark
Run this notebook on the authors’ Windows 11 study computer (10th-generation Core i5, 8 GB RAM, NVIDIA GTX 1650). The Python stages are timed automatically. Use a stopwatch for the Kinovea/manual stages and enter seconds below. Do not report results produced on another computer.


```python
from pathlib import Path
PACKAGE_ROOT=Path('..').resolve()
FRAMES=PACKAGE_ROOT/'benchmark_inputs/1001_Right_10min_scored.csv'
DATA=PACKAGE_ROOT/'data/WMSD_analysis_deidentified.xlsx'
OUTPUT=PACKAGE_ROOT/'results/computational_benchmark'

# Enter stopwatch times in seconds; leave as None until measured.
VIDEO_LOADING_SECONDS=None
KINOVEA_TRACKING_SECONDS=None
MANUAL_CORRECTION_SECONDS=None
ANGLE_EXPORT_SECONDS=None
```

```python
import subprocess, sys
cmd=[sys.executable,str(PACKAGE_ROOT/'src/run_computational_benchmark.py'),'--frames',str(FRAMES),'--data',str(DATA),'--output',str(OUTPUT)]
for flag,value in [('--video-loading-seconds',VIDEO_LOADING_SECONDS),('--kinovea-tracking-seconds',KINOVEA_TRACKING_SECONDS),('--manual-correction-seconds',MANUAL_CORRECTION_SECONDS),('--angle-export-seconds',ANGLE_EXPORT_SECONDS)]:
    if value is not None: cmd += [flag,str(value)]
print('Running:', ' '.join(cmd))
subprocess.run(cmd,check=True)
```
