# Diagnostic learning-curve analysis

This notebook evaluates how the selected combined-feature GBT behaves as the number of outer-training records increases. The modal GBT configuration from the completed inner-tuning runs is held fixed because the smallest training subsets cannot support another stable three-fold tuning loop. This is a diagnostic stability analysis and is not used for model selection or the primary performance estimate.

```python
from pathlib import Path
import subprocess
import sys

ROOT = Path.cwd()
if ROOT.name == 'notebooks':
    ROOT = ROOT.parent
command = [
    sys.executable, str(ROOT / 'src' / 'run_learning_curve.py'),
    '--data', str(ROOT / 'data' / 'WMSD_analysis_deidentified.xlsx'),
    '--output', str(ROOT / 'results' / 'learning_curve'),
    '--repeats', '10',
]
subprocess.run(command, check=True)
```
