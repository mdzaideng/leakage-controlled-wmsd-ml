# 00 — Data and leakage audit
Run this notebook before modelling. It validates the deidentified 60-record analysis file and reports exact bilateral RULA duplicates.


```python
from pathlib import Path
import pandas as pd
DATA = Path('../data/WMSD_analysis_deidentified.xlsx')
df = pd.read_excel(DATA)
df.columns = df.columns.str.strip()
df.shape, df['Risk Class'].value_counts().to_dict()
```

```python
assert len(df) == 60
assert not df['Participant ID'].duplicated().any()
assert df['Risk Class'].value_counts().to_dict() == {'Medium': 37, 'High': 17, 'Low': 6}
prohibited = {'Name','MSDs Index','Neck Pain','Shoulder Pain','Upper Back Pain','Lower Back Pain','Elbow Pain','Wrist/Hand Pain','Hips/Thighs Pain','Knees Pain','Ankles/Feet Pain'}
assert prohibited.isdisjoint(df.columns)
print('Core integrity and leakage checks: PASS')
```

```python
pairs = (df.groupby(['RULA_Left','RULA_Right'])['Participant ID'].agg(list).reset_index())
pairs = pairs[pairs['Participant ID'].str.len() > 1]
print(f'Exact duplicated bilateral RULA pairs: {len(pairs)}')
pairs
```
