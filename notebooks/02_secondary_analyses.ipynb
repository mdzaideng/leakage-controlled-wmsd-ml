# 02 — Secondary and sensitivity analyses
Produces bootstrap uncertainty estimates, sex-stratified exploratory metrics, lifestyle-variable sensitivity analyses, and class-weighting sensitivity.


```python
from pathlib import Path
import subprocess, sys
PACKAGE_ROOT = Path('..').resolve()
DATA = PACKAGE_ROOT / 'data/WMSD_analysis_deidentified.xlsx'
PRIMARY = PACKAGE_ROOT / 'results/primary/out_of_fold_predictions.csv'
OUTPUT = PACKAGE_ROOT / 'results/secondary'
REPEATS = 10
cmd = [sys.executable, str(PACKAGE_ROOT/'src/run_secondary_analysis.py'), '--data', str(DATA), '--primary-predictions', str(PRIMARY), '--output', str(OUTPUT), '--repeats', str(REPEATS)]
print('Running:', ' '.join(cmd))
subprocess.run(cmd, check=True)
```
